## Section: 1.1 - Define measures that **do not** include the cost information
- (e.g., confusion matrices, accuracy, precision, recall, F-measures, etc.).
- Consider using: from sklearn.metrics import classification_report, confusion_matrix
***

```python
# Evaluate a classification model without considering costs
def evaluate_model(y_true, y_pred):
    # Overall percentage of correct predictions
    accuracy = accuracy_score(y_true, y_pred)
    print("Accuracy:", round(accuracy, 4))

    # Precision, recall, and F1-score for each class
    print("\nClassification Report:")
    print(classification_report(y_true, y_pred, zero_division=0))

    # Compare actual classes to predicted classes
    cm = confusion_matrix(y_true, y_pred)
    print("Confusion Matrix:")
    print(cm)

    return cm
```

- Consider creating a function that takes a confusion matrix and calculates the cost.
***

```python
# # Here is an old example, with five possible values:
# def calculate_cost(conf_matrix):
# # Fill in the cost matrix values
# #                           PREDICTED VALUES
# #                      0     1     2     3     4
#     cost_matrix = [[1234, 1234, 1234, 1234, 1234],   # 0
#                    [1234, 1234, 1234, 1234, 1234],   # 1
#                    [1234, 1234, 1234, 1234, 1234],   # 2  TRUE VALUES
#                    [1234, 1234, 1234, 1234, 1234],   # 3
#                    [1234, 1234, 1234, 1234, 1234]]   # 4
#     total = 0
#     for r in range(0, 5):
#         for c in range(0, 5):
#             total = total + cost_matrix[r][c] * conf_matrix[r][c]
#     return total

def calculate_cost(conf_matrix):
    # Rows are actual values, columns are predicted values
    # Order: satisfied (0), dissatisfied (1)
    cost_matrix = np.array([
        [0, 75],
        [450, 367.50]
    ])

    # Calculate the total expected cost
    total_cost = np.sum(conf_matrix * cost_matrix)

    # Average cost per passenger
    avg_cost = total_cost / np.sum(conf_matrix)

    # Estimate yearly cost for 100,000 bookings
    yearly_cost = avg_cost * 100000

    print("Total Cost: $", round(total_cost, 2))
    print("Average Cost per Customer: $", round(avg_cost, 2))
    print("Estimated Yearly Cost: $", round(yearly_cost, 2))

    return total_cost
```

```python
# Remove unnecessary and redundant attributes
columns_to_remove = ["id", "Arrival Delay in Minutes"]

data_df = data_df.drop(columns=columns_to_remove, errors="ignore")

# Service rating attributes from Homework 1
satisfaction_columns = [
    "Seat comfort",
    "Departure/Arrival time convenient",
    "Food and drink",
    "Gate location",
    "Inflight wifi service",
    "Inflight entertainment",
    "Online support",
    "Ease of Online booking",
    "On-board service",
    "Leg room service",
    "Baggage handling",
    "Checkin service",
    "Cleanliness",
    "Online boarding"
]

# Combine the service ratings into an average
data_df["Average Satisfaction Rating"] = data_df[satisfaction_columns].mean(axis=1)

# Check the updated dataset
print("Number of attributes:", data_df.shape[1])
print("\nRemaining attributes:")
print(data_df.columns.tolist())

print("\nFirst five rows:")
display(data_df.head())
```

***
## Section: 2.3 - Transform the attributes
- Consider transforming the remaining attributes (e.g., using the data dictionary to replace the numbers with text values for some attributes – this might or might not be useful), normalizing / scaling values, encoding labels (if necessary), etc.
- Describe what you chose to do (and not do), and why.

  - You also may need to do some specific transformations below for each of the classification models you choose.
- **IMPORTANT: Any transformations you do to the datasets (particularly the Test dataset) must not artificially impact the evaluation measures.  We want the chosen classification model to work "in the real world", and the Test dataset is an approximation of the real world.**

```python
# Normalize and encode the target; remove rows with missing or unexpected labels
target_labels = data_df["satisfaction_v2"].astype("string").str.strip().str.lower()
y = target_labels.map({
    "satisfied": 0,
    "neutral or dissatisfied": 1
})

invalid_target = y.isna()
if invalid_target.any():
    print("Rows excluded due to missing or unexpected target labels:")
    print(target_labels[invalid_target].value_counts(dropna=False))

X = data_df.loc[~invalid_target].drop(columns=["satisfaction_v2"])
y = y.loc[~invalid_target]

# Split into training and testing datasets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.33, random_state=42, stratify=y
)

# Convert categorical attributes into numerical columns
categorical_columns = ["Gender", "Customer Type", "Type of Travel", "Class"]

X_train = pd.get_dummies(X_train, columns=categorical_columns, dtype=int)
X_test = pd.get_dummies(X_test, columns=categorical_columns, dtype=int)

# Make sure both datasets have the same columns
X_test = X_test.reindex(columns=X_train.columns, fill_value=0)

# Standardize the numerical attributes
numerical_columns = X_train.select_dtypes(include=np.number).columns

scaler = StandardScaler()

X_train[numerical_columns] = scaler.fit_transform(X_train[numerical_columns])
X_test[numerical_columns] = scaler.transform(X_test[numerical_columns])

# Check the results
print("Training dataset:", X_train.shape)
print("Testing dataset:", X_test.shape)

print("\nTransformed attributes:")
display(X_train.head())
```

***
## Section: 3.1 - Configure the off-the-shelf KNN classifier
- Use the KNeighborsClassifier from the SciKit Learn library

```python
# Set up the KNN classifier with 5 neighbors
knn_model = KNeighborsClassifier(
    n_neighbors=5,
    weights="uniform",
    metric="euclidean",
    algorithm="auto",
    n_jobs=-1
)

# Impute missing values using training-set medians, then apply them to both sets
train_medians = X_train.median()
X_train = X_train.fillna(train_medians)
X_test = X_test.fillna(train_medians)

# Train the classifier using the training dataset
knn_model.fit(X_train, y_train)

print("KNN classifier trained successfully.")
print("Number of neighbors:", knn_model.n_neighbors)
print("Distance metric:", knn_model.metric)
print("Weighting method:", knn_model.weights)
```

***
## Section: 3.2 - Run and evaluate the classifier
- Try several values of the K parameter and compare the results.
- Evaluate the performance of the classifier, using the evaluation methods you defined above.
***

```python
# Create a validation set from the training data
X_fit, X_val, y_fit, y_val = train_test_split(
    X_train, y_train, test_size=0.25,
    random_state=42, stratify=y_train
)

# Test several K values
k_values = [3, 4, 5, 6, 7, 11, 21]
results = []

for k in k_values:
    # Train KNN using the current K value
    knn = KNeighborsClassifier(
        n_neighbors=k,
        weights="uniform",
        metric="euclidean",
        n_jobs=-1
    )

    knn.fit(X_fit, y_fit)
    y_pred = knn.predict(X_val)

    # Calculate performance and cost
    accuracy = accuracy_score(y_val, y_pred)
    cm = confusion_matrix(y_val, y_pred, labels=[0, 1])

    cost_matrix = np.array([[0, 75], [450, 367.50]])
    total_cost = np.sum(cm * cost_matrix)

    results.append([k, accuracy, total_cost])

# Compare the results
results_df = pd.DataFrame(
    results, columns=["K", "Accuracy", "Total Cost"]
)

display(results_df)
```

***
## Section: 3.3 - Evaluate the choice of the KNN classifier
- What characteristics of the problem and data made KNN a good or bad choice?
***

```python
# Plot accuracy for each K value
plt.figure(figsize=(8, 5))

plt.plot(results_df["K"], results_df["Accuracy"], marker="o")
plt.xlabel("Number of Neighbors (K)")
plt.ylabel("Validation Accuracy")
plt.title("KNN Accuracy for Different K Values")
plt.xticks(k_values)
plt.grid(True)

plt.show()

# Plot financial cost for each K value
plt.figure(figsize=(8, 5))

plt.plot(results_df["K"], results_df["Total Cost"], marker="o")
plt.xlabel("Number of Neighbors (K)")
plt.ylabel("Total Expected Cost ($)")
plt.title("KNN Expected Cost for Different K Values")
plt.xticks(k_values)
plt.grid(True)

plt.show()
```

```python
# Select the K value with the lowest validation cost
best_k = int(results_df.loc[results_df["Total Cost"].idxmin(), "K"])

print("Best K value:", best_k)

# Train the final KNN model using the full training dataset
knn_model = KNeighborsClassifier(
    n_neighbors=best_k,
    weights="uniform",
    metric="euclidean",
    n_jobs=-1
)

knn_model.fit(X_train, y_train)

# Predict customer satisfaction on the test dataset
y_pred = knn_model.predict(X_test)

# Evaluate using the methods from Sections 1.1 and 1.2
cm = evaluate_model(y_test, y_pred)

total_cost = calculate_cost(cm)
```

```python
# Try different regularization strengths
c_values = [0.01, 0.1, 1, 10, 100]
lr_results = []

# Create a validation dataset
X_fit, X_val, y_fit, y_val = train_test_split(
    X_train, y_train, test_size=0.25,
    random_state=42, stratify=y_train
)

for c in c_values:
    # Train logistic regression with the current C value
    lr = LogisticRegression(
        C=c,
        max_iter=1000,
        solver="lbfgs",
        random_state=42
    )

    lr.fit(X_fit, y_fit)
    y_pred = lr.predict(X_val)

    # Calculate accuracy and financial cost
    accuracy = accuracy_score(y_val, y_pred)
    cm = confusion_matrix(y_val, y_pred, labels=[0, 1])

    cost_matrix = np.array([[0, 75], [450, 367.50]])
    total_cost = np.sum(cm * cost_matrix)

    lr_results.append([c, accuracy, total_cost])

# Compare the different models
lr_results_df = pd.DataFrame(
    lr_results, columns=["C", "Accuracy", "Total Cost"]
)

display(lr_results_df)
```

```python
# Try different variance smoothing values
smoothing_values = [1e-12, 1e-9, 1e-7, 1e-5, 1e-3]
nb_results = []

# Create a validation dataset
X_fit, X_val, y_fit, y_val = train_test_split(
    X_train, y_train, test_size=0.25,
    random_state=42, stratify=y_train
)

for smoothing in smoothing_values:
    # Train Naive Bayes using the current smoothing value
    nb = GaussianNB(var_smoothing=smoothing)

    nb.fit(X_fit, y_fit)
    y_pred = nb.predict(X_val)

    # Calculate accuracy and financial cost
    accuracy = accuracy_score(y_val, y_pred)
    cm = confusion_matrix(y_val, y_pred, labels=[0, 1])

    cost_matrix = np.array([[0, 75], [450, 367.50]])
    total_cost = np.sum(cm * cost_matrix)

    nb_results.append([smoothing, accuracy, total_cost])

# Compare the different models
nb_results_df = pd.DataFrame(
    nb_results, columns=["Smoothing", "Accuracy", "Total Cost"]
)

display(nb_results_df)
```
